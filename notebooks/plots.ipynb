# Single Simulation

```julia
param = SimulationParameter(
    action0=0.1f0,
    norm0=2.0f0,
    ext_pun0=0.1f0,
    int_pun_ext0=0.0f0,
    int_pun_self0=0.0f0,
    population_size=50,
    group_size=10,
    relatedness=0.5
);
```

```julia
@time sim_dict = run_simulation(param, num_replicates = 4, save_file = false);
```

```julia
sim_list = collect(values(sim_dict))
```

```julia
plot_simulation_Plots(sim_list[1], :generation, display_plot = true);
```

```julia
plot_simulation_Plotly(sim_list[1], :generation)
```
